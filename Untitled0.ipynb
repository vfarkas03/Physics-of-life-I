```python
#exercise 2

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

N = 1e9
alpha1 = 0.025
alpha2 = 0.03
T1 = 10
T2 = 60
n1_0 = 500
n2_0 = 500

#Modell für 2 bakterienstämme
def competition(t, y):
  n1, n2 = y

  if t < T1:
    growth1 = 0
  else:
    growth1 = alpha1

  if t < T2:
    growth2 = 0
  else:
    growth2 = alpha2

  dn1dt = growth1 * (1 - (n1 + n2) / N) * n1
  dn2dt = growth2 * (1 - (n1 + n2) / N) * n2

  return[dn1dt, dn2dt]


```
